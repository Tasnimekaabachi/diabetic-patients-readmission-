```python
## 1. Business Understanding

### 1.1 Contexte et problématique
Aux États-Unis, une part importante des patients diabétiques hospitalisés est **réadmise dans les 30 jours** suivant la sortie. C'est coûteux, mauvais pour le patient, et les programmes de pénalités hospitalières (type **HRRP**) sanctionnent les établissements dont les taux de réadmission sont trop élevés. Dans nos données, **11,2 %** des séjours sont suivis d'une réadmission à 30 jours.

**Problématique :** *peut-on, à partir des informations disponibles au moment de la sortie, repérer les patients à risque de réadmission à 30 jours, comprendre quels profils de patients existent, et orienter les ressources de suivi (appels, consultations rapides, éducation thérapeutique) vers les bons patients ?*

### 1.2 Les 6 objectifs métier et leur traduction en Data Science

| Objectif métier | Idée | Question | Objectif Data Science | Technique |
|---|---|---|---|---|
| **BO1** Radar de sortie | score de risque en **feu tricolore** avant la signature du bulletin | Qui risque de revenir ? | prédire `readmitted_30` | classification |
| **BO2** Personas | profils types avec un nom parlant (« séjour express », « revenant »…) | Quels types de patients ? | segmenter | ACP + K-Means |
| **BO3** Signaux faibles | traduire le modèle en messages compréhensibles par les médecins | Pourquoi reviennent-ils ? | importance des variables | permutation, taux par palier |
| **BO4** Bon patient, bonne ressource | concentrer les appels sur les patients où l'impact est maximal | Qui suivre en priorité ? | seuil de décision selon la capacité | recall / précision / lift |
| **BO5** Tableau de bord qualité | indicateurs par profil, âge, destination, pour la direction | Comment piloter ? | indicateurs agrégés | agrégations |
| **BO6** IA de confiance | modèle explicable et équitable | Le modèle est-il juste ? | audit par sous-groupes | métriques par groupe |

### 1.3 Critères de succès
- **Utile :** le score capte nettement plus de réadmissions qu'un ciblage au hasard, pour un nombre raisonnable de patients suivis.
- **Compréhensible :** personas et facteurs de risque expliqués en langage médical.
- **Juste :** pas d'écart marqué de performance entre sous-groupes (race, genre, âge).
- **Réaliste :** la littérature sur ce jeu de données rapporte un pouvoir prédictif limité (**AUC ≈ 0,6 à 0,7**). Un AUC de 0,95 serait un signal de **fuite de données**, pas une réussite.

### 1.4 Choix de la cible
`readmitted_30 = 1` si `readmitted == "<30"`, sinon 0 (`">30"` et `"NO"` sont regroupés). C'est la définition des programmes de pénalités type HRRP, et elle donne un objectif binaire clair. **Limite :** on perd la distinction « réadmis après 30 jours » (31 % des séjours), cliniquement différente de « jamais réadmis ».
```
